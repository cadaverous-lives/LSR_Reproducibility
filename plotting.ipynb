```julia
using Plots, DelimitedFiles
using LaTeXStrings
using ResearchTools
theme(:pacoty)

save = true

if save
    pgfplotsx()
else
    gr()
end
```

```julia
# alpha study
alphas = readdlm("data/alpha_study/alphas.txt")
all_res_16_alphas = readdlm("data/alpha_study/residuals_16.txt")
all_res_64_alphas = readdlm("data/alpha_study/residuals_64.txt")
all_res_256_alphas = readdlm("data/alpha_study/residuals_256.txt")
niters = size(all_res_16_alphas, 1) - 1

resplt = plot(; yaxis=:log10, legend=:topright, 
                xlabel="iteration", ylabel=L"\|r\|_{(2, \infty)}", 
                ylim=(1e-7, 1e-2), yticks=[1e-16, 1e-14, 1e-12, 1e-10, 1e-8, 1e-6, 1e-4, 1e-2, 1e0], 
                size=(600,300)
            )
plot!(resplt, 0:niters, all_res_16_alphas; label=alphas' .|> α->L"\alpha = %$α", color=[1:8...]')
plot!(resplt, 0:niters, all_res_64_alphas; label=:none, color=[1:8...]', line=:dash)
plot!(resplt, 0:niters, all_res_256_alphas; label=:none, color=[1:8...]', line=:dot)

if save
    savefig(resplt, "../figures/alpha_study.pdf")
else
    plot!()
end
# plot!(resplt, 0:niters, all_residuals_64; label=:none, color=[1:8...]', linestyle=:dash, )
```

```julia
# Nl study
all_res_16_alphas = readdlm("data/nl_study/residuals_16.txt")
all_res_64_alphas = readdlm("data/nl_study/residuals_64.txt")
all_res_256_alphas = readdlm("data/nl_study/residuals_256.txt")
niters = size(all_res_16_alphas, 1)-1
nlevels = 1:2:7

resplt = plot(; yaxis=:log10, legend=:topright, 
                xlabel="iteration", ylabel=L"\|r\|_{(2, \infty)}", 
                ylim=(1e-7, 1e1), yticks=[1e-16, 1e-14, 1e-12, 1e-10, 1e-8, 1e-6, 1e-4, 1e-2, 1e0], 
                size=(600,300)
            )
plot!(resplt, 0:niters, all_res_16_alphas; label=nlevels' .|> nl->L"N_{\ell} = %$nl", color=[1:8...]')
plot!(resplt, 0:niters, all_res_64_alphas; label=:none, color=[1:8...]', line=:dash)
plot!(resplt, 0:niters, all_res_256_alphas; label=:none, color=[1:8...]', line=:dot)

if save
    savefig(resplt, "../figures/nl_study.pdf")
else
    plot!()
end
# plot!(resplt, 0:niters, all_residuals_64; label=:none, color=[1:8...]', linestyle=:dash, )
```

```julia
# FMG study
# gr()
all_res_16 = readdlm("data/fmg_study/residuals_16.txt")
all_res_64 = readdlm("data/fmg_study/residuals_64.txt")
all_res_256 = readdlm("data/fmg_study/residuals_256.txt")
niters = size(all_res_16, 1)-1

resplt = plot(; yaxis=:log10, legend=:topright, 
                xlabel="iteration", ylabel=L"\|r\|_{(2, \infty)}", 
                ylim=(1e-15, 1e1), yticks=[1e-16, 1e-14, 1e-12, 1e-10, 1e-8, 1e-6, 1e-4, 1e-2, 1e0], 
                xlim=(-1, 21),
                size=(300,300)
            )
plot!(resplt, 0:niters, all_res_16; label=L"T_f = 16T_\lambda")
plot!(resplt, 0:niters, all_res_64; label=L"T_f = 64T_\lambda", color=1, linestyle=:dash)
plot!(resplt, 0:niters, all_res_256; label=L"T_f = 256T_\lambda", color=1, linestyle=:dot)

u0 = readdlm("data/fmg_study/init_16.txt")
uf = readdlm("data/fmg_study/final_16.txt")

traj_plot = plot(; axis=:none, ticks=:none,
                   size=(300, 300), legend=:bottomleft
                )
plot!(traj_plot, eachcol(uf[1:2:end, :])..., linewidth=0.75, label="final trajectory")
plot!(traj_plot, eachcol(u0[1:2^6:end, :])..., linewidth=0.75, alpha=0.9, label="initial guess")
lims = (xlims(traj_plot), ylims(traj_plot), zlims(traj_plot))

if save
    savefig(resplt, "../figures/fmg_study.pdf")
    savefig(traj_plot, "../figures/fmg_study_traj.pdf")
else
    plot(resplt, traj_plot; size=(600, 300))
end
```
